### 5. Video Capture & Notebook-Specific Optimization
This block implements live video streaming using `cv2.VideoCapture`. I refactored this code away from the standard command-line structure provided in the OpenCV documentation to make it fully compatible with an interactive Jupyter Notebook environment.

**Modifications in the Code:**
* **Bypassing the Jupyter Argument Trap:** The official documentation utilizes `sys.argv` to parse video sources dynamically from a terminal execution line. However, when running cells inside a Jupyter Notebook, the backend kernel passes its own hidden parameters (like `-f` and local connection paths) to sync the workspace. The original code mistook these backend flags for a video file path, causing it to crash silently. I hardcoded `s = 0` to directly route execution to the default integrated webcam.
* **Kernel Stability & Window Cleanup:** Instead of using `cv2.destroyWindow()`, I implemented `cv2.destroyAllWindows()`. Jupyter kernels are prone to caching graphical windows in system memory; using the global cleanup function ensures that all visual processes are cleanly terminated when the loop breaks, preventing the notebook from freezing.
* **Execution Flow:** I restructured the loop condition to a continuous `while True` block with an internal keypress evaluation, ensuring a cleaner execution sequence: frame capture $\rightarrow$ frame display $\rightarrow$ escape handling.

```python
import cv2

# Force the source to 0 (default built-in webcam)
s = 0
source = cv2.VideoCapture(s)

win_name = 'Camera Preview'
cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)

while True:
    has_frame, frame = source.read()
    if not has_frame:
        print("Error: Could not grab a frame from the camera.")
        break
        
    cv2.imshow(win_name, frame)
    
    # Wait for 1ms. If the Escape key (ASCII 27) is pressed, close the window
    if cv2.waitKey(1) == 27:
        break

# Clean up and completely release the camera resource
source.release()
cv2.destroyAllWindows()
```
