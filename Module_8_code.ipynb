### 1. Asset Retrieval & Setup
This block automates downloading and unzipping our template form (`form.jpg`) and a crookedly snapped photo of it (`scanned-form.jpg`).

```python
# Imports
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

from zipfile import ZipFile
from urllib.request import urlretrieve

%matplotlib inline
```

```python
def download_and_unzip(url, save_path):
    print(f"Downloading and extracting assests....", end="")

    # Downloading zip file using urllib package.
    urlretrieve(url, save_path)

    try:
        # Extracting zip file using the zipfile package.
        with ZipFile(save_path) as z:
            # Extract ZIP file contents in the same directory.
            z.extractall(os.path.split(save_path)[0])

        print("Done")

    except Exception as e:
        print("\nInvalid file.", e)
URL = r"https://www.dropbox.com/s/zuwnn6rqe0f4zgh/opencv_bootcamp_assets_NB8.zip?dl=1"

asset_zip_path = os.path.join(os.getcwd(), f"opencv_bootcamp_assets_NB8.zip")

# Download if assest ZIP does not exists. 
if not os.path.exists(asset_zip_path):
    download_and_unzip(URL, asset_zip_path)  
```

### 2. Loading the Target and Template Images
We load our reference template (the perfect alignment goal) and our skewed target image. 
* **Color Space Alignment:** Because OpenCV reads images in BGR format and Matplotlib displays them in RGB format, we use `cv2.cvtColor()` immediately to correct the color channels so our preview renders cleanly.

```python
# Read reference image
refFilename = "form.jpg"
print("Reading reference image:", refFilename)
im1 = cv2.imread(refFilename, cv2.IMREAD_COLOR)
im1 = cv2.cvtColor(im1, cv2.COLOR_BGR2RGB)

# Read image to be aligned
imFilename = "scanned-form.jpg"
print("Reading image to align:", imFilename)
im2 = cv2.imread(imFilename, cv2.IMREAD_COLOR)
im2 = cv2.cvtColor(im2, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=[20, 10]); 
plt.subplot(121); plt.axis('off'); plt.imshow(im1); plt.title("Original Form")
plt.subplot(122); plt.axis('off'); plt.imshow(im2); plt.title("Scanned Form")
```

### 3. Feature Detection (`cv2.ORB_create`)
To align two images, the computer needs to find unique anchor points (like corners, distinct shapes, or edges) common to both frames. We use an algorithm called **ORB** (Oriented FAST and Rotated BRIEF) because it is exceptionally fast and completely open-source.
* **Keypoints vs. Descriptors:** `orb.detectAndCompute()` gives out two things: **Keypoints** (the exact X,Y coordinates of the interesting features) and **Descriptors** (a numerical "fingerprint" describing what the neighborhood around that coordinate looks like).

```python
# Convert images to grayscale
im1_gray = cv2.cvtColor(im1, cv2.COLOR_BGR2GRAY)
im2_gray = cv2.cvtColor(im2, cv2.COLOR_BGR2GRAY)


# Detect ORB features and compute descriptors.
MAX_NUM_FEATURES = 500
orb = cv2.ORB_create(MAX_NUM_FEATURES)
keypoints1, descriptors1 = orb.detectAndCompute(im1_gray, None)
keypoints2, descriptors2 = orb.detectAndCompute(im2_gray, None)

# Display
im1_display = cv2.drawKeypoints(im1, keypoints1, outImage=np.array([]), 
                                color=(255, 0, 0), flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

im2_display = cv2.drawKeypoints(im2, keypoints2, outImage=np.array([]), 
                                color=(255, 0, 0), flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
```

```python
plt.figure(figsize=[20,10])
plt.subplot(121); plt.axis('off'); plt.imshow(im1_display); plt.title("Original Form");
plt.subplot(122); plt.axis('off'); plt.imshow(im2_display); plt.title("Scanned Form");
```

### 4. Feature Matching (`cv2.DescriptorMatcher_create`)
Once we have our feature fingerprints (descriptors) for both images, we need to pair them up. We initialize a Brute-Force Matcher that uses **Hamming Distance** (which mathematically calculates how many bits differ between two binary fingerprints).
* **Filtering Noise:** Not all matches are correct. We sort the matches by their quality score (`distance`) and keep only the top 10% (`numGoodMatches`). This strips away false pairings before calculating structural transformations.
* When creating a Brute-Force matcher, we can pass an optional boolean flag called `crossCheck=True`. If enabled, it forces the algorithm to check if point A matches point B *and* that point B matches back to point A, which drastically improves matching accuracy at the cost of slight processing time.

```python
# Match features.
matcher = cv2.DescriptorMatcher_create(cv2.DESCRIPTOR_MATCHER_BRUTEFORCE_HAMMING)

# Converting to list for sorting as tuples are immutable objects.
matches = list(matcher.match(descriptors1, descriptors2, None))

# Sort matches by score
matches.sort(key=lambda x: x.distance, reverse=False)

# Remove not so good matches
numGoodMatches = int(len(matches) * 0.1)
matches = matches[:numGoodMatches]
```

```python
# Draw top matches
im_matches = cv2.drawMatches(im1, keypoints1, im2, keypoints2, matches, None)

plt.figure(figsize=[40, 10])
plt.imshow(im_matches);plt.axis("off");plt.title("Original Form")
```

### 5. Computing the Homography Matrix (`cv2.findHomography`)
A **Homography** is a 3x3 mathematical transformation matrix ($H$) that maps coordinates from one flat plane to another, warping an arbitrary quadrilateral back into a perfect square layout:

$$x_{template} = H \cdot x_{scanned}$$

To solve this matrix equation, the computer requires a minimum of **4 corresponding point pairs**.
* **The Power of RANSAC:** Because even our top 10% of matches can contain bad pairings (outliers), we use the **RANSAC** algorithm. RANSAC acts as a statistical lie-detector: it tests random subsets of points over and over until it finds the mathematically truest layout consensus, throwing away the bad coordinates automatically.
  * `ransacReprojThreshold` (default 3.0): The maximum pixel error allowed to consider a point pair an "inlier". Lowering this forces tighter structural accuracy.
  * `maxIters` (default 2000): The maximum number of internal guessing rounds RANSAC will attempt.

```python
# Extract location of good matches
points1 = np.zeros((len(matches), 2), dtype=np.float32)
points2 = np.zeros((len(matches), 2), dtype=np.float32)

for i, match in enumerate(matches):
    points1[i, :] = keypoints1[match.queryIdx].pt
    points2[i, :] = keypoints2[match.trainIdx].pt

# Find homography
h, mask = cv2.findHomography(points2, points1, cv2.RANSAC)
```

### 6. Perspective Warping (`cv2.warpPerspective`)
With our calculated Homography matrix ($h$) in hand, we apply it to our crooked image using `cv2.warpPerspective()`. This function shifts, rotates, and stretches the pixel coordinate arrays of our scanned form to snap it perfectly flush with the original template dimensions.
* **Additional Parameters:**
  * `borderMode` and `borderValue`: If our warping leaves behind empty black pixels along the margins, we can use these flags to automatically pad the empty edges with a solid color or mirror adjacent pixels.

```python
# Use homography to warp image
height, width, channels = im1.shape
im2_reg = cv2.warpPerspective(im2, h, (width, height))

# Display results
plt.figure(figsize=[20, 10])
plt.subplot(121);plt.imshow(im1);    plt.axis("off");plt.title("Original Form")
plt.subplot(122);plt.imshow(im2_reg);plt.axis("off");plt.title("Scanned Form")
```
