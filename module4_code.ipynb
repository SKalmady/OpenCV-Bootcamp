### 1. Preprocessing and Asset Automation
This block handles the download and extraction of the asset files required for this module (the coast, window, sheet music, and logo images). 

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

from zipfile import ZipFile
from urllib.request import urlretrieve

from IPython.display import Image

%matplotlib inline
```

```python
def download_and_unzip(url, save_path):
    print(f"Downloading and extracting assests....", end="")

    # Downloading zip file using urllib package.
    urlretrieve(url, save_path)
    
    try:
        # Extracting zip file using the zipfile package.
        with ZipFile(save_path) as z:
            # Extract ZIP file contents in the same directory.
            z.extractall(os.path.split(save_path)[0])

        print("Done")

    except Exception as e:
        print("\nInvalid file.", e)

URL = r"https://www.dropbox.com/s/0oe92zziik5mwhf/opencv_bootcamp_assets_NB4.zip?dl=1"

asset_zip_path = os.path.join(os.getcwd(), f"opencv_bootcamp_assets_NB4.zip")

# Download if assest ZIP does not exists. 
if not os.path.exists(asset_zip_path):
    download_and_unzip(URL, asset_zip_path)  
```

### 2. Image Arithmetic for Brightness Adjustment
Image brightness is altered globally by adding or subtracting a constant scalar value from every pixel coordinate matrix. 
* **Mechanics:** `cv2.add()` and `cv2.subtract()` perform **saturated operations**. This means if a pixel intensity exceeds 255 or drops below 0, OpenCV automatically caps it at 255 or 0 to keep it valid for 8-bit unsigned data (`uint8`).
* **Hidden Parameters (Unused):**
  * `mask`: An optional 8-bit single-channel array. If passed, the addition/subtraction is *only* performed on the pixel coordinates where the mask contains non-zero values.
  * `dtype`: Optional output array depth (e.g., `cv2.CV_16U`), useful if we deliberately want to prevent saturation and store values in a higher bit depth.

```python
img_bgr = cv2.imread("New_Zealand_Coast.jpg", cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Display 18x18 pixel image.
Image(filename="New_Zealand_Coast.jpg")
```

```python
matrix = np.ones(img_rgb.shape, dtype="uint8") * 50
#create a matrix with the original values of the image and then multiply it with the desired number. This will be added or subtracted from the original image to vary brightness.
img_rgb_brighter = cv2.add(img_rgb, matrix)
img_rgb_darker   = cv2.subtract(img_rgb, matrix)


# Show the images
plt.figure(figsize=[18, 5])
plt.subplot(131); plt.imshow(img_rgb_darker);  plt.title("Darker");
plt.subplot(132); plt.imshow(img_rgb);         plt.title("Original");
plt.subplot(133); plt.imshow(img_rgb_brighter);plt.title("Brighter");
```

### 3. Image Multiplication, Contrast, and the Overflow Trap
While addition changes brightness, multiplication changes **contrast** (the relative scaling difference between light and dark regions). 
* **The Glitch:** Raw multiplication can cause standard integer arrays to wrap around or artifact wildly if values exceed 255.
* **The Solution:** To safely scale contrast, the data matrix must first be temporarily cast to floating-point (`np.float64`), multiplied, scaled safely using `np.clip(array, min, max)` to force values strictly between 0 and 255, and then cast back down to `np.uint8`.

```python
matrix1 = np.ones(img_rgb.shape) * 0.8
matrix2 = np.ones(img_rgb.shape) * 1.2

img_rgb_darker   = np.uint8(cv2.multiply(np.float64(img_rgb), matrix1))
img_rgb_brighter = np.uint8(cv2.multiply(np.float64(img_rgb), matrix2))

# Show the images
plt.figure(figsize=[18,5])
plt.subplot(131); plt.imshow(img_rgb_darker);  plt.title("Lower Contrast");
plt.subplot(132); plt.imshow(img_rgb);         plt.title("Original");
plt.subplot(133); plt.imshow(img_rgb_brighter);plt.title("Higher Contrast");
```

```python
matrix1 = np.ones(img_rgb.shape) * 0.8
matrix2 = np.ones(img_rgb.shape) * 1.2

img_rgb_lower  = np.uint8(cv2.multiply(np.float64(img_rgb), matrix1))
img_rgb_higher = np.uint8(np.clip(cv2.multiply(np.float64(img_rgb), matrix2), 0, 255))

# Show the images
plt.figure(figsize=[18,5])
plt.subplot(131); plt.imshow(img_rgb_lower); plt.title("Lower Contrast");
plt.subplot(132); plt.imshow(img_rgb);       plt.title("Original");
plt.subplot(133); plt.imshow(img_rgb_higher);plt.title("Higher Contrast");
```

### 4. Image Thresholding (Binarization)
Thresholding converts a grayscale image into a binary image (pure black and white), which is vital for isolating features or creating masks.



* **Mechanics:** In `cv2.threshold(src, thresh, maxval, type)`, if a pixel's value is greater than the `thresh` value, it gets mapped to `maxval`. Otherwise, it becomes 0.
* **Threshold Types:** While `cv2.THRESH_BINARY` is used here, the official documentation lists several alternatives:
  * `cv2.THRESH_BINARY_INV`: Flips the logic (pixels above threshold become black, below become white).
  * `cv2.THRESH_TRUNC`: Truncates values. Pixels above the threshold are capped at `thresh`, while pixels below remain unchanged.
  * `cv2.THRESH_TOZERO`: Pixels below the threshold become 0; pixels above remain unchanged.

```python
img_read = cv2.imread("building-windows.jpg", cv2.IMREAD_GRAYSCALE)
retval, img_thresh = cv2.threshold(img_read, 100, 255, cv2.THRESH_BINARY)

# Show the images
plt.figure(figsize=[18, 5])

plt.subplot(121);plt.imshow(img_read, cmap="gray");  plt.title("Original")
plt.subplot(122);plt.imshow(img_thresh, cmap="gray");plt.title("Thresholded")

print(img_thresh.shape)
```

### 5. Global vs. Adaptive Thresholding
Global thresholding fails completely when an image has uneven lighting or shadows (like the sheet music sample). Adaptive thresholding fixes this by calculating a localized threshold for every individual pixel based on a small neighborhood window around it.
* **Key Parameters Explored:**
  * `blockSize`: The size of the pixel neighborhood (must be an odd number like 3, 5, 11).
  * `C`: A constant value subtracted from the calculated local average to fine-tune the noise threshold.
* **Adaptive Methods:**
  * This code uses `cv2.ADAPTIVE_THRESH_MEAN_C` (simple average of the neighborhood box).
  * `cv2.ADAPTIVE_THRESH_GAUSSIAN_C` is a powerful alternative that calculates a weighted sum based on a Gaussian distribution, making the thresholding far more robust to organic gradients and shadows.

```python
# Read the original image
img_read = cv2.imread("Piano_Sheet_Music.png", cv2.IMREAD_GRAYSCALE)

# Perform global thresholding
retval, img_thresh_gbl_1 = cv2.threshold(img_read, 50, 255, cv2.THRESH_BINARY)

# Perform global thresholding
retval, img_thresh_gbl_2 = cv2.threshold(img_read, 130, 255, cv2.THRESH_BINARY)

# Perform adaptive thresholding
img_thresh_adp = cv2.adaptiveThreshold(img_read, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 7)

# Show the images
plt.figure(figsize=[18,15])
plt.subplot(221); plt.imshow(img_read,        cmap="gray");  plt.title("Original");
plt.subplot(222); plt.imshow(img_thresh_gbl_1,cmap="gray");  plt.title("Thresholded (global: 50)");
plt.subplot(223); plt.imshow(img_thresh_gbl_2,cmap="gray");  plt.title("Thresholded (global: 130)");
plt.subplot(224); plt.imshow(img_thresh_adp,  cmap="gray");  plt.title("Thresholded (adaptive)");
```

### 6. Logical Bitwise Operations
Bitwise operations analyze individual pixel values at the binary level. In image processing, black pixels act as `0` (False) and white pixels act as `255` (True). 



* **`bitwise_and`**: Output is true only if both overlapping pixels are white (finds the intersection).
* **`bitwise_or`**: Output is true if at least one pixel is white (combines the shapes).
* **`bitwise_xor`**: Output is true only if one pixel is white and the other is black (highlights the differences).
* **`bitwise_not`**: Inverts the image array values completely.

```python
img_rec = cv2.imread("rectangle.jpg", cv2.IMREAD_GRAYSCALE)

img_cir = cv2.imread("circle.jpg", cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=[20, 5])
plt.subplot(121);plt.imshow(img_rec, cmap="gray")
plt.subplot(122);plt.imshow(img_cir, cmap="gray")
print(img_rec.shape)
```

```python
result = cv2.bitwise_and(img_rec, img_cir, mask=None)
plt.imshow(result, cmap="gray")
```

```python
result = cv2.bitwise_or(img_rec, img_cir, mask=None)
plt.imshow(result, cmap="gray")
```

```python
result = cv2.bitwise_xor(img_rec, img_cir, mask=None)
plt.imshow(result, cmap="gray")
```

### 7. Complex Application: Masking and Logo Manipulation
This pipeline brings everything together to composite a new background texture into the text boundaries of a foreground logo. 

**The Pipeline Logic:**
1. **Isolate Geometry:** Convert the logo to grayscale and threshold it to create a clean, black-and-white mask of the layout text.
2. **Isolate Background:** Use the original mask to cut out a text-shaped piece of our color checkerboard background.
3. **Isolate Foreground:** Use `cv2.bitwise_not()` to invert the mask, allowing us to clip the solid red background *out* of the original logo while leaving the lettering empty.
4. **Merge Matrices:** Use simple matrix addition via `cv2.add()` to seamlessly weld the isolated background slice and foreground slice into a single finished frame.

```python
Image(filename='Logo_Manipulation.png')
```

```python
img_bgr = cv2.imread("coca-cola-logo.png")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)

print(img_rgb.shape)

logo_w = img_rgb.shape[0]
logo_h = img_rgb.shape[1]
```

```python
# Read in image of color cheackerboad background
img_background_bgr = cv2.imread("checkerboard_color.png")
img_background_rgb = cv2.cvtColor(img_background_bgr, cv2.COLOR_BGR2RGB)

# Set desired width (logo_w) and maintain image aspect ratio
aspect_ratio = logo_w / img_background_rgb.shape[1]
dim = (logo_w, int(img_background_rgb.shape[0] * aspect_ratio))

# Resize background image to sae size as logo image
img_background_rgb = cv2.resize(img_background_rgb, dim, interpolation=cv2.INTER_AREA)

plt.imshow(img_background_rgb)
print(img_background_rgb.shape)
```

```python
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Apply global thresholding to creat a binary mask of the logo
retval, img_mask = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(img_mask, cmap="gray")
print(img_mask.shape)
```

```python
# Create an inverse mask
img_mask_inv = cv2.bitwise_not(img_mask)
plt.imshow(img_mask_inv, cmap="gray")
```

```python
# Create colorful background "behind" the logo lettering
img_background = cv2.bitwise_and(img_background_rgb, img_background_rgb, mask=img_mask)
plt.imshow(img_background)
```

```python
# Isolate foreground (red from original image) using the inverse mask
img_foreground = cv2.bitwise_and(img_rgb, img_rgb, mask=img_mask_inv)
plt.imshow(img_foreground)
```

```python
# Add the two previous results obtain the final result
result = cv2.add(img_background, img_foreground)
plt.imshow(result)
cv2.imwrite("logo_final.png", result[:, :, ::-1])
```
